**Ejercicio 2:** Desarrolle un programa en que permita realizar operaciones entre vectores. El
programa debe permitir al usuario ingresar los vectores y realizar las siguientes operaciones:
 Suma de vectores

*   Producto punto
*   Producto cruz
*    Producto de Hadamard entre vectores


Debe mostrar un menú de opciones para que el usuario elija la operación que desea realizar. El
programa debe ejecutarse repetidamente hasta que el usuario elija la opción de salir. También
debe reconocer cuando el usuario ingresó caracteres por error, como cadenas.

```python
def sumar_vectores(A, B):
    n = len(A)
    if n != len(B):
        return "Error: Los vectores deben tener las mismas dimensiones para sumarse."
    else:
        try:
            resultado = []
            for i in range(n):
                resultado.append(float(A[i]) + float(B[i]))
            return resultado
        except ValueError:
            return "Error: Debe ingresar solo numeros."


def producto_punto(A, B):
    n = len(A)
    if n != len(B):
        return "Error: Los vectores deben tener las mismas dimensiones para el producto punto."
    else:
        try:
            resultado = 0
            for i in range(n):
                resultado += float(A[i]) * float(B[i])
            return resultado
        except ValueError:
            return "Error: Debe ingresar solo numeros."


def producto_cruz(A, B):
    if len(A) != 3 or len(B) != 3:
        return "Error: El producto cruz solo esta definido para vectores de 3 dimensiones."
    else:
        try:
            a0, a1, a2 = float(A[0]), float(A[1]), float(A[2])
            b0, b1, b2 = float(B[0]), float(B[1]), float(B[2])
            resultado =[
                a1 * b2 - a2 * b1,
                a2 * b0 - a0 * b2,
                a0 * b1 - a1 * b0]
            return resultado
        except ValueError:
            return "Error: Debe ingresar solo numeros."


def hadamard(A, B):
    n = len(A)
    if n != len(B):
        return "Error: Los vectores deben tener las mismas dimensiones para el producto Hadamard."
    else:
        try:
            resultado = []
            for i in range(n):
                resultado.append(float(A[i]) * float(B[i]))
            return resultado
        except ValueError:
            return "Error: Debe ingresar solo numeros."


while True:
    try:
        print("      ---OPERACIONES ENTRE VECTORES---     ")
        print("---------- ")
        print("1. Suma de vectores")
        print("2. Producto punto")
        print("3. Producto cruz")
        print("4. Producto de Hadamard")
        print("5. Salir")
        print("---------- ")


        opcion = int(input("Elige una opcion: "))

        if opcion == 1:
            v1 = input("Ingrese el vector 1 separado por espacios: ").split()
            v2 = input("Ingrese el vector 2 separado por espacios: ").split()
            r = sumar_vectores(v1, v2)
            print("La suma es:", r)

        elif opcion == 2:
            v1 = input("Ingrese el vector 1 separado por espacios: ").split()
            v2 = input("Ingrese el vector 2 separado por espacios: ").split()
            r = producto_punto(v1, v2)
            print("El producto punto es:", r)

        elif opcion == 3:
            v1 = input("Ingrese el vector 1 separado por espacios: ").split()
            v2 = input("Ingrese el vector 2 separado por espacios: ").split()
            r = producto_cruz(v1, v2)
            print("El producto cruz es:", r)

        elif opcion == 4:
            v1 = input("Ingrese el vector 1 separado por espacios: ").split()
            v2 = input("Ingrese el vector 2 separado por espacios: ").split()
            r = hadamard(v1, v2)
            print("El producto Hadamard es:", r)

        elif opcion == 5:
            print("Fin del programa")
            break

        else:
            print("Opcion incorrecta, ingrese un numero entre 1 y 5.")

    except ValueError:
        print("Opcion incorrecta, debe ingresar un numero entero.")
```

**Ejercicio 3:** Desarrolle un programa en que permita realizar operaciones entre matrices. El
programa debe permitir al usuario ingresar las matrices y realizar las siguientes operaciones:

*   Suma de matrices
*   Multiplicación de matrices
*   Producto de Hadamard
*   Producto de Kronecker

Debe mostrar un menú de opciones para que el usuario elija la operación que desea realizar. El
programa debe ejecutarse repetidamente hasta que el usuario elija la opción de salir. También
debe reconocer cuando el usuario ingresó caracteres por error, como cadenas.

```python
def ingresar_matriz():
  """Función para ingresar una matriz por teclado"""
  filas = int(input("Ingrese el número de filas: "))
  columnas = int(input("Ingrese el número de columnas: "))
  matriz = []
  for i in range(filas):
    while True:
      fila = list(map(float, input(f"Ingrese la fila {i+1}separada por espacios: ").split()))
      if len(fila) != columnas:
        print(f"Error: debe ingresar exactamente {columnas}valores.")
      else:
        matriz.append(fila)
        break
  return matriz

def suma_matrices(A, B):
  filas = len(A)
  columnas = len(A[0])

  resultado = []

  for i in range(filas):
        fila = []
        for j in range(columnas):
            fila.append(A[i][j] + B[i][j])
        resultado.append(fila)

  return resultado

def multiplicar_matrices(A, B):
  filas_A = len(A)
  columnas_A = len(A[0])
  columnas_B = len(B[0])

  resultado = [[0]*columnas_B for _ in range(filas_A)]

  for i in range(filas_A):
    for j in range(columnas_B):
      for k in range(columnas_A):
        resultado[i][j] += A[i][k] * B[k][j]

  return resultado

def hadamard(A, B):
  filas = len(A)
  columnas = len(A[0])

  resultado = []

  for i in range(filas):
    fila = []

    for j in range(columnas):
      fila.append(A[i][j] * B[i][j])
      resultado.append(fila)

  return resultado

def kronecker(A, B):
  resultado = []
  for fila_A in A:
    for fila_B in B:
      fila = []
      for elemento_A in fila_A:
        for elemento_B in fila_B:
          fila.append(elemento_A * elemento_B)
          resultado.append(fila)

  return resultado

while True:
  print("----Menú de opciones:----")
  print("1. Sumar matrices X y Z")
  print("2. Multiplicar matrices X y Z")
  print("3. Calcular Hadamard de matrices X y Z")
  print("4. Calcular Kronecker de matrices X y Z")
  print("5. Salir")

  opcion = input("Ingrese una opción: ")

  if opcion == "1":
        X = ingresar_matriz()
        Z = ingresar_matriz()
        resultado_suma = suma_matrices(X, Z)
        print("----> Resultado de la suma de matrices:", resultado_suma)

  elif opcion == "2":
        X = ingresar_matriz()
        Z = ingresar_matriz()
        resultado_multiplicacion = multiplicar_matrices(X, Z)
        print("----> Resultado de la multiplicación de matrices:", resultado_multiplicacion)

  elif opcion == "3":
        X = ingresar_matriz()
        Z = ingresar_matriz()
        resultado_hadamard = hadamard(X, Z)
        print("----> Resultado de la Hadamard de matrices:", resultado_hadamard)

  elif opcion == "4":
        X = ingresar_matriz()
        Z = ingresar_matriz()
        resultado_kronecker = kronecker(X, Z)
        print("----> Resultado de la Kronecker de matrices:", resultado_kronecker)

  elif opcion == "5":
      print("--->Saliendo del programa...")
      break

  else:
        print("Opción inválida. Por favor, ingrese una opción válida.")
```

Ejercicio 4

Desarrolle un programa que permita realizar transformaciones de matrices.  
El programa debe permitir al usuario ingresar matrices y realizar las siguientes operaciones:

- Transponer una matriz  
- Determinar la matriz de cofactores  
- Vectorizar por filas una matriz  

Sea la matriz:

A =
\begin{pmatrix}
1 & 2 \\
3 & 4
\end{pmatrix}

La vectorización por filas es:

\begin{pmatrix}
1 \\
2 \\
3 \\
4
\end{pmatrix}

Debe mostrar un menú de opciones para que el usuario elija la operación que desea realizar. El
programa debe ejecutarse repetidamente hasta que el usuario elija la opción de salir. También
debe reconocer cuando el usuario ingresó caracteres por error, como cadenas.

```python
def ingresar_matriz():
  try:
    filas = int(input("Ingrese el numero de filas: "))
    columnas = int(input("Ingrese el numero de columnas: "))

    matriz = []
    for i in range(filas):
      fila = input(f"Ingrese los elementos de la fila {i+1} separados por espacios: ").split()
      fila = [float(elemento) for elemento in fila]

      if len(fila) != columnas:
        print("El numero de columnas debe ser el mismo para todas las filas")
        return None

      matriz.append([float(x) for x in fila])

    return matriz

  except ValueError:
    print("Entrada inválida")
    return None

def transpuesta(A):
    filas = len(A)
    columnas = len(A[0])

    T = []
    for j in range(columnas):
        fila = []
        for i in range(filas):
            fila.append(A[i][j])
        T.append(fila)

    return T


def submatriz(A, fila_excluir, col_excluir):
    return [
        [A[i][j] for j in range(len(A)) if j != col_excluir]
        for i in range(len(A)) if i != fila_excluir
    ]


def determinante(A):
    if len(A) == 1:
        return A[0][0]

    if len(A) == 2:
        return A[0][0]*A[1][1] - A[0][1]*A[1][0]

    det = 0
    for j in range(len(A)):
        menor = submatriz(A, 0, j)
        det += ((-1)**j) * A[0][j] * determinante(menor)

    return det


def cofactores(A):
    n = len(A)

    for fila in A:
        if len(fila) != n:
            print(" La matriz debe ser cuadrada")
            return None

    C = []

    for i in range(n):
        fila = []
        for j in range(n):
            menor = submatriz(A, i, j)
            cofactor = ((-1)**(i+j)) * determinante(menor)
            fila.append(cofactor)
        C.append(fila)

    return C


def vectorizar_filas(A):
    vec = []
    for fila in A:
        for elemento in fila:
            vec.append([elemento])

    return vec


def imprimir_matriz(A):
    for fila in A:
        print(fila)


def menu():
    while True:
        print("\n--- MENÚ ---")
        print("1. Transponer matriz")
        print("2. Matriz de cofactores")
        print("3. Vectorizar por filas")
        print("4. Salir")

        opcion = input("Seleccione una opción: ")

        if opcion == "4":
            print("Fin del programa")
            break

        A = ingresar_matriz()
        if A is None:
            continue

        if opcion == "1":
            print("Resultado:")
            imprimir_matriz(transpuesta(A))

        elif opcion == "2":
            resultado = cofactores(A)
            if resultado:
                print("Resultado:")
                imprimir_matriz(resultado)

        elif opcion == "3":
            print("Resultado:")
            imprimir_matriz(vectorizar_filas(A))

        else:
            print("Opción no valida")


menu()
```
